```python
from sentence_transformers import SentenceTransformer, util
import torch

# Use GPU if available
device = "cuda" if torch.cuda.is_available() else "cpu"
model = SentenceTransformer("l3cube-pune/indic-sentence-bert-nli").to(device)
```

```python
def solve_analogy(A, B, C, candidate_Ds):
    # Encode A, B, C
    vec = lambda text: model.encode(text, convert_to_tensor=True, device=device)
    A_vec = vec(A)
    B_vec = vec(B)
    C_vec = vec(C)

    # Compute analogy vector: B - A + C
    analogy_vector = B_vec - A_vec + C_vec

    # Get embeddings for all candidate Ds
    D_embeddings = model.encode(candidate_Ds, convert_to_tensor=True, device=device)

    # Compute cosine similarities
    scores = util.cos_sim(analogy_vector, D_embeddings)
    best_idx = torch.argmax(scores).item()

    return candidate_Ds[best_idx], scores[0][best_idx].item()
```

```python
analogy_data = [
    {"A": "அண்ணன்", "B": "ஆண்", "C": "அக்கா", "D": "பெண்"},
    {"A": "தமிழ்", "B": "மொழி", "C": "இந்தி", "D": "மொழி"},
    {"A": "நகர்", "B": "மக்கள்", "C": "பள்ளி", "D": "மாணவர்கள்"},
]

candidate_Ds = ["பெண்", "ஆண்", "மொழி", "மாணவர்கள்", "ஆசிரியர்கள்", "நகரம்"]
```

```python
for row in analogy_data:
    pred, score = solve_analogy(row["A"], row["B"], row["C"], candidate_Ds)
    print(f"{row['A']} : {row['B']} :: {row['C']} : {pred}  ({row['D']})  Score: {score:.2f}")
```

```python
correct = 0
for row in analogy_data:
    pred, _ = solve_analogy(row["A"], row["B"], row["C"], candidate_Ds)
    if pred == row["D"]:
        correct += 1
accuracy = correct / len(analogy_data)
print(f"Accuracy: {accuracy * 100:.2f}%")
```
