```python
from transformers import pipeline
classifier = pipeline('sentiment-analysis')
```

```python
res = classifier(['He was a good boy','so maybe now also'])
for result in res:
  print(f"label:{result['label']},with score:{round(result['score'])}")
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
```

```python
model_name="nlptown/bert-base-multilingual-uncased-sentiment"
model=AutoModelForSequenceClassification.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)
classifier = pipeline('sentiment-analysis',model=model,tokenizer = tokenizer)
```

```python
import pandas as pd
df=pd.read_csv('df_file.csv',sep=',',
               names=["label","message"],
               header=0)
df.head()
```

```python
df.shape
```

```python
X=list(df['message'].astype(str).fillna(''))
```

```python
y=list(df['label'])
```

```python
y = pd.get_dummies(y, drop_first=True).values.tolist()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=0)
```
